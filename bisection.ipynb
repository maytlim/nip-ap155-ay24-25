```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def some_log(x):
    return np.log(x)
```

```python
def plot_me(f, xmin=0.1, xmax=5, npts=100):
    x = np.linspace(xmin, xmax, npts)
    plt.plot(x, f(x))
```

```python
#let me draw what I'm trying to solve first
xmin, xmax = 0.1, 5
plot_me(some_log, xmin, xmax)
```

```python
def get_midpoint(x_left, x_right):
    return (x_left + x_right)/2
```

```python
def bisect_once(x_left, x_right, f):
    x_mid = get_midpoint(x_left, x_right)
    if f(x_left) * f(x_mid) <= 0:
        x_right = x_mid
    else:
        x_left = x_mid
    return x_left, x_right
```

```python
def check_bisect(x0, x1, f):
    if (f(x0) * f(x1)) <= 0:
        return True
    else:
        return False
```

```python
def bisection_method(f, x_left=0.1, x_right=4, delta=1e-8):
    root = None
    sigma = (x_right - x_left)/2
    if check_bisect(x_left, x_right, f):
        while (sigma/get_midpoint(x_left, x_right)) > delta:
            x_left, x_right = bisect_once(x_left, x_right, f)
            root = get_midpoint(x_left, x_right)
            sigma = (x_right - x_left)/2
    return root, sigma
```

```python
root, sigma = bisection_method(some_log, x_left=0.2, x_right=4, delta=1e-8)
print('Using the bisection method, the root is at x= {0:8e} +/- {1:8e}'.format(root, sigma))
```
